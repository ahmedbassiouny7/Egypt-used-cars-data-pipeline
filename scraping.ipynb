```python
import os
import time
import random
from datetime import datetime

import pandas as pd

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
```

```python
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized")

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
)
print("Browser ready!")
```

```python
driver.get("https://eg.hatla2ee.com/en/car/page/1")
time.sleep(3)

#grid_xpath = "/html/body/div[@class='container']/div[@class='grid grid-cols-1 lg:pt-5 gap-4 lg:gap-6']/div[@class='grid grid-cols-1 lg:grid-cols-[268px_1fr] gap-5']/div[@class='flex flex-col gap-6 min-w-0']/div[@class='flex flex-col gap-6']/div[@class='grid grid-cols-[repeat(auto-fill,minmax(230px,1fr))] auto-rows-max gap-4 lg:gap-y-6']"

cards = driver.find_elements(
    By.XPATH,
    "//div[contains(@class, 'auto-rows-max')]/span/div[contains(@class, 'card')]"
)

print(len(cards))
print(cards[0].text)
```
